```python
import numpy as np
```

```python


X = df[df["origin"] == "Asia"][["vehicle_weight", "model_year"]].head(7).to_numpy()

#Multiply the transpose of X with X
XTX = X.T @ X

#Calculate the inverse of XTX
inverse_XTX = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

#Calculate the w
w = inverse_XTX @ X.T @ y

#sum w
print(w.sum())
```
