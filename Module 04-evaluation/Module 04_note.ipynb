

## Module Overview: Evaluation Metrics for Binary Classification

This module covers how to rigorously measure, interpret, and compare the performance of binary classification models (continuing with the telco customer churn prediction project).

### Key Objectives & Workflow:

* **Accuracy & Baselines:** Evaluate basic model accuracy and understand why dummy models (e.g., predicting "no churn" for everyone) serve as essential sanity checks.
* **Confusion Table:** Break down correct predictions and errors into a four-way confusion table (true positives, true negatives, false positives, false negatives).
* **Precision & Recall:** Explore specialized metrics that focus strictly on the positive class to evaluate trade-offs between false alarms and missed detections.
* **ROC Curves:** Analyze True Positive Rates (TPR) and False Positive Rates (FPR) across all possible classification decision thresholds against random and ideal model baselines.
* **ROC AUC:** Quantify overall curve quality using the Area Under the ROC Curve (AUC) as a single robust metric.
* **Cross-Validation:** Implement cross-validation for more reliable model quality estimation and hyperparameter tuning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline 
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('/workspaces/ML-Zoomcamp/Module 03-classification/data-week-3.csv')

df.columns = df.columns.str.lower().str.replace(' ', '_')

categorical_columns = list(df.dtypes[df.dtypes == 'string'].index)

for c in categorical_columns:
    df[c] = df[c].str.lower().str.replace(' ', '_')

df.totalcharges = pd.to_numeric(df.totalcharges, errors='coerce')
df.totalcharges = df.totalcharges.fillna(0)

df.churn = (df.churn == 'yes').astype(int)
```

```python
df_full_train, df_test = train_test_split(df, test_size= 0.2, random_state= 1)
df_train, df_val = train_test_split(df_full_train, test_size= 0.25, random_state= 1)

df_train = df_train.reset_index(drop = True)
df_val = df_val.reset_index(drop = True)
df_test = df_test.reset_index(drop = True)

y_train = df_train.churn.values
y_val = df_val.churn.values
y_test = df_test.churn.values

del df_train['churn']
del df_val['churn']
del df_test['churn']
```

**ROC Curves**

```python
tpr = tp / (tp + fn)
fpr = fp / (fp + tn)
```

```python
# TPR and FPR at all thresholds
scores = []

thresholds = np.linspace(0, 1, 101)

for t in thresholds:
    actual_positive = (y_val == 1)
    actual_negative = (y_val == 0)
    
    predict_positive = (y_pred_val >= t)
    predict_negative = (y_pred_val < t)

    tp = (predict_positive & actual_positive).sum()
    tn = (predict_negative & actual_negative).sum()

    fp = (predict_positive & actual_negative).sum()
    fn = (predict_negative & actual_positive).sum()
    
    scores.append((t, tp, fp, fn, tn))

columns = ['threshold', 'tp', 'fp', 'fn', 'tn']
df_scores = pd.DataFrame(scores, columns=columns)

df_scores['tpr'] = df_scores.tp / (df_scores.tp + df_scores.fn)
df_scores['fpr'] = df_scores.fp / (df_scores.fp + df_scores.tn)
```

```python
plt.plot(df_scores.threshold, df_scores['tpr'], label='TPR')
plt.plot(df_scores.threshold, df_scores['fpr'], label='FPR')
plt.legend()
```

```python
def tpr_fpr_dataframe(y_val, y_pred):
    scores = []

    thresholds = np.linspace(0, 1, 101)

    for t in thresholds:
        actual_positive = (y_val == 1)
        actual_negative = (y_val == 0)
    
        predict_positive = (y_pred >= t)
        predict_negative = (y_pred < t)

        tp = (predict_positive & actual_positive).sum()
        tn = (predict_negative & actual_negative).sum()

        fp = (predict_positive & actual_negative).sum()
        fn = (predict_negative & actual_positive).sum()
    
        scores.append((t, tp, fp, fn, tn))

    columns = ['threshold', 'tp', 'fp', 'fn', 'tn']
    df_scores = pd.DataFrame(scores, columns=columns)

    df_scores['tpr'] = df_scores.tp / (df_scores.tp + df_scores.fn)
    df_scores['fpr'] = df_scores.fp / (df_scores.fp + df_scores.tn)

    return df_scores
```

**Cross-Validation**

```python
def train(df_train, y_train, C=1.0):
    dicts = df_train[categorical + numerical].to_dict(orient='records')

    dv = DictVectorizer(sparse=False)
    X_train = dv.fit_transform(dicts)

    model = LogisticRegression(C=C, max_iter=1000)
    model.fit(X_train, y_train)
    
    return dv, model
```

```python
def predict(df, dv, model):
    dicts = df[categorical + numerical].to_dict(orient='records')

    X = dv.transform(dicts)
    y_pred_val = model.predict_proba(X)[:, 1]

    return y_pred_val
```

```python
from sklearn.model_selection import KFold
```

```python
kfold = KFold(n_splits=5, shuffle=True, random_state=1)
```

```python

from tqdm.auto import tqdm
import warnings
from sklearn.exceptions import ConvergenceWarning

warnings.filterwarnings('ignore', category=ConvergenceWarning)
```

```python

n_splits = 5

for C in tqdm([0.001, 0.01, 0.1, 0.5, 1, 5, 10]):
    kfold = KFold(n_splits=n_splits, shuffle=True, random_state=1)

    scores = []

    for train_idx, val_idx in kfold.split(df_full_train):
        df_train = df_full_train.iloc[train_idx]
        df_val = df_full_train.iloc[val_idx]

        y_train = df_train.churn.values
        y_val = df_val.churn.values

        dv, model = train(df_train, y_train, C=C)
        y_pred_val = predict(df_val, dv, model)

        auc = roc_auc_score(y_val, y_pred_val)
        scores.append(auc)

    print('C=%s %.3f +- %.3f' % (C, np.mean(scores), np.std(scores)))
```

```python
#Training the final model
dv, model = train(df_full_train, df_full_train.churn.values, C=1.0)
y_pred_test = predict(df_test, dv, model)

auc = roc_auc_score(y_test, y_pred_test)
auc
```

## **The metrics in one place**
**Metric**: a single number that describes the performance of a model

**Accuracy**: the fraction of correct answers; sometimes misleading, especially with class imbalance - a dummy model predicting "no churn" for everybody already gets 73%

**Precision and recall**: less misleading when we have class imbalance; precision checks how many of our positive predictions are right, recall checks how many real positives we caught

**ROC Curve**: a way to evaluate the performance at all thresholds; okay to use with imbalance

**ROC AUC**: the area under the ROC curve - 0.5 for a random model, 1.0 for an ideal one; interpretable as the probability that a random positive scores higher than a random negative

**K-Fold CV**: a more reliable estimate for performance (mean + std), used for parameter tuning
